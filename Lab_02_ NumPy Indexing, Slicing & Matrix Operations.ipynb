Task 1: Apply Indexing and Slicing

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Retail_Dataset_Modified.csv")
print("Dataset Loaded Succesfully")
```

```python
revenue_array = np.array(df["Revenue"])
print(revenue_array)
```

```python
category_array = np.array(df["Category"])
print(category_array)
```

```python
store_name_array = np.array(df["Store_Name"])
print(store_name_array)
```

```python
store_type_array = np.array(df["Store_Type"])
print(store_type_array)
```

```python
city_array = np.array(df["City"])
print(city_array)
```

```python
print("First Revenue Value:")
print(revenue_array[0])
```

```python
print("Last Revenue Value:")
print(revenue_array[-1])
```

```python
print("First Five Revenue Values:")
print(revenue_array[0:5])
```

```python
print("Revenue Values from Index 10 to 20:")
print(revenue_array[10:21])
```

```python
sales_array =  np.array(df[["Units_Sold", "Profit_Margin"]])
print(sales_array)
```

```python
print("First Row:")
print(sales_array[0])
```

```python
print("Profit Margin Column:")
print(sales_array[:,1])
```
