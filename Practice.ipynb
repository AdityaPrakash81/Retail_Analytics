Array performs math, list dont

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print(list1 + list2)

import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print(arr1 + arr2)

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print(arr2 - arr1)

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print(arr1 * arr2)

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print(arr2 / arr1)
```

```python
import numpy as np 
batsman_a = np.array([48, 50, 49, 51, 51, 50])
batsman_b = np.array([10, 90, 30, 70, 100, 0])

print("A -> Variance:", np.var(batsman_a), "Std Dev:", np.std(batsman_a))
print("b -> Variance:", np.var(batsman_b), "Std Dev:", np.std(batsman_b))
```
